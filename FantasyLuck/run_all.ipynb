# Running everything

```python
import time, nbformat
from pathlib import Path
from nbclient import NotebookClient
from nbclient.exceptions import CellExecutionError
from IPython.display import Pretty, display

NOTEBOOKS = sorted(Path(".").glob("0[1-6]*.ipynb"))
print("Running:")
for p in NOTEBOOKS:
    print("  " + p.name)
```

```python
def printed_text(nb):
    """Everything the notebook printed, so the test results can be checked here."""
    lines = []
    for cell in nb.cells:
        for out in cell.get("outputs", []):
            if out.get("output_type") == "stream":
                lines += out.get("text", "").splitlines()
    return lines

def clock(seconds):
    m, s = divmod(int(seconds), 60)
    return f"{m}:{s:02d}"

def bar(done, total, width=30):
    filled = round(width * done / total) if total else width
    return "█" * filled + "░" * (width - filled)

# Count the code cells first, so the bar can show progress across all six notebooks
books = [(path, nbformat.read(path, as_version=4)) for path in NOTEBOOKS]
total_cells = sum(c.cell_type == "code" for _, nb in books for c in nb.cells)
done_cells = 0

status = display(Pretty("Starting…"), display_id=True)   # one line that updates in place
start_all = time.time()
for n, (path, nb) in enumerate(books, 1):
    start = time.time()
    code_cells = [i for i, c in enumerate(nb.cells) if c.cell_type == "code"]
    client = NotebookClient(nb, timeout=None, kernel_name="python3",
                            resources={"metadata": {"path": str(path.parent.resolve())}})
    try:
        with client.setup_kernel():
            for k, i in enumerate(code_cells, 1):
                status.update(Pretty(f"{bar(done_cells, total_cells)}  {done_cells / total_cells:4.0%}   "
                              f"{path.name} ({n} of {len(books)}), cell {k} of {len(code_cells)}   "
                              f"{clock(time.time() - start_all)} so far"))
                client.execute_cell(nb.cells[i], i)
                done_cells += 1
    except CellExecutionError:
        nbformat.write(nb, path)   # saves the outputs up to the error, so you can open it and look
        status.update(Pretty(f"STOPPED in {path.name}, cell {k} of {len(code_cells)}, after {clock(time.time() - start_all)}"))
        print(f"\nSTOPPED in {path.name}. The error is below, and the notebook is saved with its outputs.")
        raise
    nbformat.write(nb, path)       # saves every cell's output back into the notebook
    print(f"Finished {path.name} in {clock(time.time() - start)}")

    # Show the test results each notebook printed
    for line in printed_text(nb):
        if line.lstrip().startswith("Test") or "don't add up" in line or " off:" in line or "Saved" in line:
            print("    " + line.strip())

status.update(Pretty(f"{bar(1, 1)}  100%   All {len(books)} notebooks ran in {clock(time.time() - start_all)}"))
```
